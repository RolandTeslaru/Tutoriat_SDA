## c1 - complexitate

de cate ori se executa `a += 1` si `b += 1`? ce complexitate au?

```python
import math


def bucle(n):
    a = 0
    for i in range(n):
        for j in range(i, n):
            a += 1
    b = 0
    i = 1
    while i < n:
        i *= 2
        b += 1
    return a, b


for n in [10, 100, 1000]:
    a, b = bucle(n)
    print(f"n = {n}: a = {a} (n(n+1)/2 = {n * (n + 1) // 2}), b = {b} (log2 n = {math.log2(n):.1f})")
```

parantezele sunt corecte? `([]{()})`, `([)]`, `((`, `())`

```python
def corect(s):
    pereche = {")": "(", "]": "[", "}": "{"}
    st = []
    for c in s:
        if c in "([{":
            st.append(c)
        elif not st or st.pop() != pereche[c]:
            return False
    return not st


for s in ["([]{()})", "([)]", "((", "())"]:
    print(s, corect(s))
```

## c3 - arbore binar de cautare

inseram pe rand 8 3 10 1 6 14 4 7 13, apoi parcurgerile si inaltimea

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


def insert(node, key):
    if node is None:
        return Node(key)
    if key < node.key:
        node.left = insert(node.left, key)
    else:
        node.right = insert(node.right, key)
    return node


def preordine(node):
    return [node.key] + preordine(node.left) + preordine(node.right) if node else []


def inordine(node):
    return inordine(node.left) + [node.key] + inordine(node.right) if node else []


def postordine(node):
    return postordine(node.left) + postordine(node.right) + [node.key] if node else []


# in muchii, un singur nod are inaltimea 0
def inaltime(node):
    return -1 if node is None else 1 + max(inaltime(node.left), inaltime(node.right))


# arborele culcat: dreapta e sus, stanga e jos
def afiseaza(node, depth=0):
    if node:
        afiseaza(node.right, depth + 1)
        print("    " * depth + str(node.key))
        afiseaza(node.left, depth + 1)


root = None
for x in [8, 3, 10, 1, 6, 14, 4, 7, 13]:
    root = insert(root, x)

afiseaza(root)
print("preordine: ", preordine(root))
print("inordine:  ", inordine(root))  # iese sortat
print("postordine:", postordine(root))
print("inaltime:", inaltime(root))
```

## c4 - min-heap

inseram 6 4 5 1 2 8 7 3, apoi scoatem minimul de 2 ori

```python
def sift_up(h, i):
    while i > 0 and h[i] < h[(i - 1) // 2]:
        p = (i - 1) // 2
        h[i], h[p] = h[p], h[i]
        i = p


def sift_down(h, i):
    while True:
        st, dr, m = 2 * i + 1, 2 * i + 2, i
        if st < len(h) and h[st] < h[m]:
            m = st
        if dr < len(h) and h[dr] < h[m]:
            m = dr
        if m == i:
            return
        h[i], h[m] = h[m], h[i]
        i = m


h = []
for x in [6, 4, 5, 1, 2, 8, 7, 3]:
    h.append(x)
    sift_up(h, len(h) - 1)
    print("inserez", x, "->", h)

for _ in range(2):
    mn = h[0]
    h[0] = h[-1]
    h.pop()
    sift_down(h, 0)
    print("scot", mn, "->", h)
```

## c5 - mergesort

interclasare [1, 4, 7, 9] cu [2, 3, 8, 10, 11], cate comparatii?

```python
# minim min(n, m) comparatii, maxim n + m - 1
def merge(a, b, show=False):
    res, i, j, comp = [], 0, 0, 0
    while i < len(a) and j < len(b):
        comp += 1
        if show:
            print(f"compar {a[i]} cu {b[j]}", end=" -> ")
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
        if show:
            print("iau", res[-1])
    # restul se copiaza fara comparatii
    res += a[i:] + b[j:]
    return res, comp


res, comp = merge([1, 4, 7, 9], [2, 3, 8, 10, 11], show=True)
print(res, comp, "comparatii")
```

mergesort pe [38, 27, 43, 3, 9, 82, 10]

```python
def mergesort(v, depth=0):
    if len(v) <= 1:
        return v
    mid = len(v) // 2
    print("   " * depth + f"impart {v} -> {v[:mid]} {v[mid:]}")
    left = mergesort(v[:mid], depth + 1)
    right = mergesort(v[mid:], depth + 1)
    res, _ = merge(left, right)
    print("   " * depth + f"interclasez {left} {right} -> {res}")
    return res


mergesort([38, 27, 43, 3, 9, 82, 10])
```

inversiuni (i < j cu v[i] > v[j]): cate are [2, 4, 1, 3, 5]?

```python
# cu mergesort: cand iau din dreapta, face inversiune cu tot ce a ramas in stanga
def inversiuni(v):
    if len(v) <= 1:
        return v, 0
    mid = len(v) // 2
    a, x = inversiuni(v[:mid])
    b, y = inversiuni(v[mid:])
    res, i, j, inv = [], 0, 0, x + y
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else:
            inv += len(a) - i
            res.append(b[j])
            j += 1
    res += a[i:] + b[j:]
    return res, inv


print(inversiuni([2, 4, 1, 3, 5])[1])  # (2,1) (4,1) (4,3)
print(inversiuni([5, 4, 3, 2, 1])[1])  # descrescator => n(n-1)/2
```
