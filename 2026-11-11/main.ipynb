## 1. countsort

sortam [4, 1, 3, 4, 2, 0, 1, 4] numarand aparitiile

```python
v = [4, 1, 3, 4, 2, 0, 1, 4]
k = max(v) + 1
f = [0] * k
for x in v:
    f[x] += 1
print("frecvente:    ", f)

for i in range(1, k):
    f[i] += f[i - 1]
print("sume partiale:", f)  # f[x] = cate elemente sunt <= x

out = [None] * len(v)
for x in reversed(v):
    out[f[x] - 1] = x
    f[x] -= 1
    print(f"pun {x} pe pozitia {f[x]}: {out}")
```

## 2. stabilitate

sortam studentii dupa nota. la note egale ramane ordinea initiala?

```python
# key(x) = cheia dupa care sortam, intre 0 si k-1
def countsort(a, key, k, de_la_final=True):
    f = [0] * k
    for x in a:
        f[key(x)] += 1
    for i in range(1, k):
        f[i] += f[i - 1]
    out = [None] * len(a)
    for x in (reversed(a) if de_la_final else a):
        out[f[key(x)] - 1] = x
        f[key(x)] -= 1
    return out


studenti = [("Ana", 9), ("Bogdan", 7), ("Cristi", 9), ("Dana", 10), ("Elena", 7), ("Florin", 9)]
print(countsort(studenti, lambda s: s[1], 11))
# de la inceput: la egalitate ordinea se inverseaza => nu e stabil
print(countsort(studenti, lambda s: s[1], 11, de_la_final=False))
```

## 3. radixsort

countsort pe fiecare cifra, de la unitati spre cifra cea mai semnificativa. [170, 45, 75, 90, 802, 24, 2, 66]

```python
def radixsort(a, de_la_final=True):
    exp = 1
    while max(a) // exp > 0:
        a = countsort(a, lambda x: (x // exp) % 10, 10, de_la_final)
        print(f"cifra {exp:>3}: {a}")
        exp *= 10
    return a


radixsort([170, 45, 75, 90, 802, 24, 2, 66])
```

## 4. de ce trebuie countsort stabil?

acelasi radixsort, dar cu countsort de la inceput

```python
radixsort([170, 45, 75, 90, 802, 24, 2, 66], de_la_final=False)  # gresit
```

## 5. cand nu merge countsort
