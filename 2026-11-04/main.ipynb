## 1. partitionare lomuto

pivotul e ultimul element. cum arata [5, 3, 8, 4, 2, 7, 1, 6] dupa partitionare?

```python
def lomuto(a, lo, hi, show=False):
    p = a[hi]
    i = lo
    for j in range(lo, hi):
        if a[j] < p:
            a[i], a[j] = a[j], a[i]
            if show:
                print(f"{a[i]} < {p} -> il mut pe pozitia {i}: {a}")
            i += 1
    a[i], a[hi] = a[hi], a[i]
    if show:
        print(f"pivotul {p} pe pozitia {i}: {a}")
    return i


a = [5, 3, 8, 4, 2, 7, 1, 6]
lomuto(a, 0, len(a) - 1, show=True)
```

## 2. partitionare hoare

pivotul e elementul din mijloc, acelasi vector

```python
def hoare(a, lo, hi, show=False):
    p = a[(lo + hi) // 2]
    i, j = lo - 1, hi + 1
    while True:
        i += 1
        while a[i] < p:
            i += 1
        j -= 1
        while a[j] > p:
            j -= 1
        if i >= j:
            if show:
                print(f"stop la j = {j}: {a[lo:j + 1]} {a[j + 1:hi + 1]}")
            return j
        a[i], a[j] = a[j], a[i]
        if show:
            print(f"swap a[{i}] cu a[{j}]: {a}")


a = [5, 3, 8, 4, 2, 7, 1, 6]
hoare(a, 0, len(a) - 1, show=True)
```

## 3. quicksort

[38, 27, 43, 3, 9, 82, 10], cu lomuto (pivotul intre paranteze drepte)

```python
def quicksort(a, lo, hi, depth=0):
    if lo >= hi:
        return
    k = lomuto(a, lo, hi)
    print("   " * depth + f"{a[lo:k]} [{a[k]}] {a[k + 1:hi + 1]}")
    quicksort(a, lo, k - 1, depth + 1)
    quicksort(a, k + 1, hi, depth + 1)


a = [38, 27, 43, 3, 9, 82, 10]
quicksort(a, 0, len(a) - 1)
print(a)
```

```python
# cu hoare: recursiv pe [lo, j] si [j+1, hi], pivotul nu e neaparat pe pozitia j
def quicksort_hoare(a, lo, hi):
    if lo < hi:
        j = hoare(a, lo, hi)
        quicksort_hoare(a, lo, j)
        quicksort_hoare(a, j + 1, hi)


a = [38, 27, 43, 3, 9, 82, 10]
quicksort_hoare(a, 0, len(a) - 1)
print(a)
```

## 4. cazul rau

vector deja sortat + pivotul ultimul element => O(n^2). cu pivot aleator scapam de asta

```python
import random

comp = 0


def quicksort_count(a, lo, hi, aleator=False):
    global comp
    if lo >= hi:
        return
    if aleator:
        r = random.randint(lo, hi)
        a[r], a[hi] = a[hi], a[r]
    comp += hi - lo
    k = lomuto(a, lo, hi)
    quicksort_count(a, lo, k - 1, aleator)
    quicksort_count(a, k + 1, hi, aleator)


n = 500
for aleator in [False, True]:
    comp = 0
    quicksort_count(list(range(n)), 0, n - 1, aleator)
    print("pivot aleator:" if aleator else "pivot ultimul:", comp, "comparatii")
print("n^2 / 2 =", n * n // 2)
```

## 5. quickselect

al k-lea cel mai mic element fara sa sortam tot vectorul. care e mediana lui [7, 10, 4, 3, 20, 15, 8, 1, 12]?

```python
# O(n) in medie, pt ca mergem doar intr-o parte: n + n/2 + n/4 + ... <= 2n
def quickselect(v, k):
    a = list(v)
    lo, hi = 0, len(a) - 1
    while True:
        r = random.randint(lo, hi)
        a[r], a[hi] = a[hi], a[r]
        pos = lomuto(a, lo, hi)
        if pos == k - 1:
            return a[pos]
        if k - 1 < pos:
            hi = pos - 1
        else:
            lo = pos + 1


v = [7, 10, 4, 3, 20, 15, 8, 1, 12]
print(sorted(v))
print("mediana:", quickselect(v, (len(v) + 1) // 2))
```
