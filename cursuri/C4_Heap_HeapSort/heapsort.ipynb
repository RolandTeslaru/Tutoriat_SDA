# HeapSort

1. Construim un **max-heap** din vector – O(n).
2. Repetăm de n − 1 ori: interschimbăm rădăcina (maximul) cu ultimul element din heap, micșorăm heap-ul cu 1 și coborâm noua rădăcină – O(log n).

Proprietăți:
- complexitate **O(n log n)** în toate cazurile
- **in-place** (memorie suplimentară O(1))
- **nu este stabil**

```python
def sift_down(a, i, n):
    """Coboară a[i] în heap-ul a[0:n]."""
    while True:
        left, right, largest = 2 * i + 1, 2 * i + 2, i
        if left < n and a[left] > a[largest]:
            largest = left
        if right < n and a[right] > a[largest]:
            largest = right
        if largest == i:
            return
        a[i], a[largest] = a[largest], a[i]
        i = largest


def heapsort(a):
    n = len(a)
    for i in range(n // 2 - 1, -1, -1):   # build heap
        sift_down(a, i, n)
    for end in range(n - 1, 0, -1):
        a[0], a[end] = a[end], a[0]       # maximul la locul lui final
        sift_down(a, 0, end)
    return a
```

## Urmărire pas cu pas

```python
def heapsort_verbose(a):
    a = list(a)
    n = len(a)
    for i in range(n // 2 - 1, -1, -1):
        sift_down(a, i, n)
    print(f"după build heap:  {a}")
    for end in range(n - 1, 0, -1):
        a[0], a[end] = a[end], a[0]
        sift_down(a, 0, end)
        print(f"heap {str(a[:end]):28} | sortat {a[end:]}")
    return a


heapsort_verbose([4, 10, 3, 5, 1, 8, 7])
```

## Testare

```python
import random

for _ in range(1000):
    arr = [random.randint(-50, 50) for _ in range(random.randint(0, 30))]
    assert heapsort(list(arr)) == sorted(arr)
print("toate testele au trecut")
```

## De ce nu este stabil?

Elementele egale își pot schimba ordinea relativă. Sortăm perechi după prima componentă și urmărim a doua.

```python
class Item:
    def __init__(self, key, label):
        self.key, self.label = key, label
    def __gt__(self, other):
        return self.key > other.key
    def __repr__(self):
        return f"{self.key}{self.label}"


arr = [Item(2, "a"), Item(1, "x"), Item(2, "b"), Item(2, "c")]
print("înainte:", arr)
print("după:   ", heapsort(arr))   # cele trei 2-uri nu mai sunt în ordinea a, b, c
```

## Comparație de timp

```python
import time

for n in [10_000, 100_000, 200_000]:
    arr = [random.random() for _ in range(n)]
    t = time.perf_counter()
    heapsort(list(arr))
    t_heap = time.perf_counter() - t
    t = time.perf_counter()
    sorted(arr)
    t_sorted = time.perf_counter() - t
    print(f"n={n:>7}: heapsort {t_heap:.3f}s | sorted() {t_sorted:.4f}s")
```
