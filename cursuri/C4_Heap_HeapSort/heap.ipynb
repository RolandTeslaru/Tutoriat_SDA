# Heap (ansamblu)

Un **max-heap** este un arbore binar **complet** în care fiecare nod are cheia ≥ cheile fiilor. Deci maximul este în rădăcină.

Fiind complet, îl reținem compact într-un **vector** (indexare de la 0):
- parent(i) = (i − 1) // 2
- left(i) = 2i + 1
- right(i) = 2i + 2

| Operație | Complexitate |
|---|---|
| maxim | O(1) |
| inserare (urcare / *sift up*) | O(log n) |
| extragere maxim (coborâre / *sift down*) | O(log n) |
| construire din vector (*build heap*) | **O(n)** |

```python
class MaxHeap:
    def __init__(self, items=None):
        self.a = list(items) if items else []
        self._build_heap()

    def __len__(self):
        return len(self.a)

    def max(self):
        if not self.a:
            raise IndexError("heap gol")
        return self.a[0]

    def insert(self, x):
        self.a.append(x)
        self._sift_up(len(self.a) - 1)

    def extract_max(self):
        if not self.a:
            raise IndexError("heap gol")
        top = self.a[0]
        last = self.a.pop()
        if self.a:
            self.a[0] = last
            self._sift_down(0)
        return top

    def _sift_up(self, i):
        while i > 0:
            parent = (i - 1) // 2
            if self.a[i] <= self.a[parent]:
                break
            self.a[i], self.a[parent] = self.a[parent], self.a[i]
            i = parent

    def _sift_down(self, i):
        n = len(self.a)
        while True:
            left, right, largest = 2 * i + 1, 2 * i + 2, i
            if left < n and self.a[left] > self.a[largest]:
                largest = left
            if right < n and self.a[right] > self.a[largest]:
                largest = right
            if largest == i:
                break
            self.a[i], self.a[largest] = self.a[largest], self.a[i]
            i = largest

    def _build_heap(self):
        # frunzele sunt deja heap-uri; coborâm nodurile interne, de la ultimul spre rădăcină
        for i in range(len(self.a) // 2 - 1, -1, -1):
            self._sift_down(i)

    def is_heap(self):
        return all(self.a[(i - 1) // 2] >= self.a[i] for i in range(1, len(self.a)))
```

```python
h = MaxHeap()
for x in [5, 3, 17, 10, 84, 19, 6, 22, 9]:
    h.insert(x)
    print(f"insert {x:2} -> {h.a}")

print("\neste heap?", h.is_heap())
print("extrageri:", [h.extract_max() for _ in range(len(h))])
```

## Construire în O(n) (*build heap*)

De ce O(n) și nu O(n log n)? Jumătate din noduri sunt frunze (cost 0), un sfert coboară cel mult 1 nivel, o optime cel mult 2 niveluri...

$$\sum_{h=0}^{\log n} \frac{n}{2^{h+1}} \cdot h = O(n)$$

```python
arr = [4, 1, 3, 2, 16, 9, 10, 14, 8, 7]   # exemplul din Cormen
h = MaxHeap(arr)
print("vector inițial: ", arr)
print("după build heap:", h.a, "| este heap?", h.is_heap())
```

## Afișare ca arbore

```python
def print_heap(a, i=0, depth=0):
    if i >= len(a):
        return
    print_heap(a, 2 * i + 2, depth + 1)
    print("    " * depth + str(a[i]))
    print_heap(a, 2 * i + 1, depth + 1)


print_heap(h.a)
```

## Aplicație: coadă de priorități și cele mai mari k elemente

În Python există modulul `heapq`, care implementează un **min-heap**. Pentru max-heap se folosește trucul cu chei negate.

Cele mai mari `k` elemente dintr-un șir de `n`: ținem un min-heap de dimensiune `k` → **O(n log k)**.

```python
import heapq
import random


def top_k(items, k):
    h = []
    for x in items:
        if len(h) < k:
            heapq.heappush(h, x)
        elif x > h[0]:            # h[0] = cel mai mic dintre cele mai mari k
            heapq.heapreplace(h, x)
    return sorted(h, reverse=True)


data = [random.randint(1, 1000) for _ in range(20)]
print(data)
print("top 5:     ", top_k(data, 5))
print("verificare:", sorted(data, reverse=True)[:5])
```
