# Arbori binari de căutare (ABC / BST)

Un arbore binar în care, pentru orice nod `x`:
- toate cheile din subarborele **stâng** sunt **< x.key**
- toate cheile din subarborele **drept** sunt **> x.key**

Consecință: **parcurgerea în inordine dă cheile sortate**.

Căutare, inserare, ștergere: **O(h)**, unde `h` = înălțimea.
- arbore echilibrat: h ≈ log n
- cel mai rău caz (chei inserate sortat): h = n − 1 → arborele degenerează într-o listă

```python
class Node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    # ---------- căutare ----------
    def search(self, key):
        node = self.root
        while node is not None and node.key != key:
            node = node.left if key < node.key else node.right
        return node

    # ---------- inserare ----------
    def insert(self, key):
        self.root = self._insert(self.root, key)

    def _insert(self, node, key):
        if node is None:
            return Node(key)
        if key < node.key:
            node.left = self._insert(node.left, key)
        elif key > node.key:
            node.right = self._insert(node.right, key)
        return node   # cheile duplicate sunt ignorate

    # ---------- minim / maxim ----------
    @staticmethod
    def minimum(node):
        while node.left is not None:
            node = node.left
        return node

    @staticmethod
    def maximum(node):
        while node.right is not None:
            node = node.right
        return node

    # ---------- ștergere ----------
    def delete(self, key):
        self.root = self._delete(self.root, key)

    def _delete(self, node, key):
        if node is None:
            return None
        if key < node.key:
            node.left = self._delete(node.left, key)
        elif key > node.key:
            node.right = self._delete(node.right, key)
        else:
            # cazurile 1 și 2: 0 sau 1 fiu -> îl înlocuim cu fiul
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left
            # cazul 3: 2 fii -> copiem succesorul (minimul din dreapta) și îl ștergem pe acela
            succ = self.minimum(node.right)
            node.key = succ.key
            node.right = self._delete(node.right, succ.key)
        return node

    # ---------- succesor ----------
    def successor(self, key):
        """Cea mai mică cheie strict mai mare decât `key`."""
        node, candidate = self.root, None
        while node is not None:
            if key < node.key:
                candidate = node
                node = node.left
            else:
                node = node.right
        return candidate.key if candidate else None

    # ---------- utilitare ----------
    def inorder(self):
        result = []
        def visit(node):
            if node:
                visit(node.left)
                result.append(node.key)
                visit(node.right)
        visit(self.root)
        return result

    def height(self):
        def h(node):
            return -1 if node is None else 1 + max(h(node.left), h(node.right))
        return h(self.root)

    def print_tree(self):
        def show(node, depth):
            if node:
                show(node.right, depth + 1)
                print("    " * depth + str(node.key))
                show(node.left, depth + 1)
        show(self.root, 0)
```

```python
t = BST()
for x in [50, 30, 70, 20, 40, 60, 80, 35, 45]:
    t.insert(x)

t.print_tree()
print("inordine:", t.inorder())
print("min:", BST.minimum(t.root).key, "| max:", BST.maximum(t.root).key)
print("caută 45:", t.search(45) is not None, "| caută 55:", t.search(55) is not None)
print("succesorul lui 45:", t.successor(45))
```

## Cele 3 cazuri de ștergere

```python
t.delete(20)   # frunză
print("după ștergerea lui 20 (frunză):  ", t.inorder())
t.delete(30)   # 2 fii -> înlocuit cu succesorul 35
print("după ștergerea lui 30 (2 fii):   ", t.inorder())
t.delete(70)   # 2 fii
print("după ștergerea lui 70 (2 fii):   ", t.inorder())
print()
t.print_tree()
```

## Cel mai rău caz: chei inserate în ordine sortată

```python
import random

n = 500
sorted_tree, random_tree = BST(), BST()
keys = list(range(n))
for x in keys:
    sorted_tree.insert(x)
random.shuffle(keys)
for x in keys:
    random_tree.insert(x)

print(f"n = {n}")
print("înălțime cu chei sortate:  ", sorted_tree.height())   # n - 1
print("înălțime cu chei aleatoare:", random_tree.height())   # ~ c * log n
```
