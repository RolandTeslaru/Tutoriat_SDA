# Arbori binari

Un **arbore binar** este fie vid, fie format dintr-o rădăcină cu un subarbore stâng și unul drept (ambii arbori binari).

Terminologie:
- **rădăcină** – nodul fără părinte
- **frunză** – nod fără fii
- **adâncimea** unui nod – numărul de muchii de la rădăcină la el
- **înălțimea** arborelui – adâncimea maximă (arbore cu un singur nod: 0)
- arbore **strict** – fiecare nod are 0 sau 2 fii
- arbore **complet** – toate nivelurile pline, cu excepția ultimului, completat de la stânga

```python
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


#         1
#        / \
#       2   3
#      / \   \
#     4   5   6
#        /
#       7
root = Node(1,
            Node(2, Node(4), Node(5, Node(7))),
            Node(3, None, Node(6)))
```

## 1. Parcurgeri în adâncime (recursive)

- **Preordine** (RSD): rădăcină, stânga, dreapta
- **Inordine** (SRD): stânga, rădăcină, dreapta
- **Postordine** (SDR): stânga, dreapta, rădăcină

Toate sunt O(n).

```python
def preorder(node):
    if node is None:
        return []
    return [node.value] + preorder(node.left) + preorder(node.right)


def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.value] + inorder(node.right)


def postorder(node):
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.value]


print("preordine: ", preorder(root))
print("inordine:  ", inorder(root))
print("postordine:", postorder(root))
```

## 2. Parcurgere pe niveluri (BFS) – folosește o **coadă**

```python
from collections import deque


def level_order(root):
    if root is None:
        return []
    levels, q = [], deque([root])
    while q:
        level = []
        for _ in range(len(q)):
            node = q.popleft()
            level.append(node.value)
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        levels.append(level)
    return levels


print(level_order(root))
```

## 3. Inordine iterativă – folosește o **stivă**

Mergem cât putem la stânga punând nodurile pe stivă; când nu mai putem, scoatem un nod, îl vizităm și trecem în dreapta lui.

```python
def inorder_iterative(root):
    result, stack, node = [], [], root
    while stack or node:
        while node:
            stack.append(node)
            node = node.left
        node = stack.pop()
        result.append(node.value)
        node = node.right
    return result


print(inorder_iterative(root))
```

## 4. Funcții recursive pe arbori

Schema generală: rezolvăm pentru subarborele stâng și cel drept, apoi combinăm.

```python
def count_nodes(node):
    if node is None:
        return 0
    return 1 + count_nodes(node.left) + count_nodes(node.right)


def count_leaves(node):
    if node is None:
        return 0
    if node.left is None and node.right is None:
        return 1
    return count_leaves(node.left) + count_leaves(node.right)


def height(node):
    if node is None:
        return -1          # arbore vid; un singur nod are înălțimea 0
    return 1 + max(height(node.left), height(node.right))


def mirror(node):
    if node is None:
        return None
    return Node(node.value, mirror(node.right), mirror(node.left))


print("noduri:", count_nodes(root))
print("frunze:", count_leaves(root))
print("înălțime:", height(root))
print("oglindit, pe niveluri:", level_order(mirror(root)))
```

## 5. Reconstrucție din preordine + inordine

Primul element din preordine e rădăcina. Poziția ei în inordine împarte elementele în subarborele stâng și cel drept.
(Funcționează dacă valorile sunt distincte.)

```python
def build_tree(pre, ino):
    if not pre:
        return None
    r = pre[0]
    k = ino.index(r)
    return Node(r,
                build_tree(pre[1:k + 1], ino[:k]),
                build_tree(pre[k + 1:], ino[k + 1:]))


rebuilt = build_tree(preorder(root), inorder(root))
print(postorder(rebuilt) == postorder(root))
```

## 6. Afișare „culcată” a arborelui (util la tablă / debug)

```python
def print_tree(node, depth=0):
    if node is None:
        return
    print_tree(node.right, depth + 1)
    print("    " * depth + str(node.value))
    print_tree(node.left, depth + 1)


print_tree(root)
```
